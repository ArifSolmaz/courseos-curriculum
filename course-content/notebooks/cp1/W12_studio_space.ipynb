## Data Exploration

Before moving to the standard target, let us explore your data.

```python
# Explore the structure of your data
print("=== Data Exploration for Lightcurve Explorer ===")
print()

# What columns do we have?
if sample_data:
    print("Columns:", list(sample_data[0].keys()))
    print()

# Look at the flux column specifically
print("flux values:")
for i, row in enumerate(sample_data):
    val = row.get("flux", "N/A")
    print(f"  Row {i}: flux = {val} (type: {type(val).__name__})")

print()
# Quick stats on numeric values
numeric_vals = []
for row in sample_data:
    try:
        numeric_vals.append(float(row["flux"]))
    except (ValueError, TypeError, KeyError):
        pass

if numeric_vals:
    print(f"Numeric flux values: {len(numeric_vals)} of {len(sample_data)}")
    print(f"  Min: {min(numeric_vals)}")
    print(f"  Max: {max(numeric_vals)}")
    print(f"  Mean: {sum(numeric_vals)/len(numeric_vals):.2f}")
```
