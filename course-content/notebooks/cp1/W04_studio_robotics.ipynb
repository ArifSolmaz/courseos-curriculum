### Step 1: Set up config

```python
config = {
    "project_name": "mechasense_studio",
    "track": "robotics",
    "version": "v1",
    "value_column": "temp",
    "threshold": 60,
    "min_value": 0,
    "max_value": 150,
}
```

### Step 2: Implement clean_data() for temp

```python
def clean_data(data, config):
    """Clean Robotics/Mechatronics data."""
    cleaned = []
    dropped = 0

    for row in data:
        val = row.get("temp")

        # Rule: skip missing/invalid values
        if val is None or val == "":
            dropped += 1
            continue

        try:
            num = float(val)
        except (ValueError, TypeError):
            dropped += 1
            continue

        # Rule: must be in range 0 to 150
        if num < 0 or num > 150:
            dropped += 1
            continue

        cleaned.append({**row, "temp": num})

    print(f"Cleaned: {len(data)} -> {len(cleaned)} ({dropped} dropped)")
    return cleaned

result = clean_data(sample_data, config)
print()
print("Clean rows:")
for row in result:
    print(f"  {row}")

assert len(result) < len(sample_data), "Should have dropped at least 1 row"
print()
print("Must-pass core: PASSED!")
```
