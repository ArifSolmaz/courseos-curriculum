## Part 1: True or False?

Python can answer yes/no questions using **comparisons**:

````python
# Comparisons return True or False
print(10 > 5)     # True
print(10 < 5)     # False
print(10 == 10)   # True  (== means "equal to")
print(10 != 5)    # True  (!= means "not equal to")
print(10 >= 10)   # True  (greater than or equal)
print(10 <= 9)    # False (less than or equal)
````

**Common mistake:** `=` vs `==`
- `x = 5` means "assign 5 to x"
- `x == 5` means "is x equal to 5?"

````python
score = 73

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score} → Grade: {grade}")
````

## Part 3: Combining Conditions

Use `and`, `or`, `not` to combine conditions:

````python
temp = 25
humidity = 80

# Both must be true
if temp > 20 and humidity > 70:
    print("It's warm AND humid")

# At least one must be true
if temp > 30 or humidity > 90:
    print("Extreme condition!")
else:
    print("Conditions are moderate")

# Flip a condition
is_broken = False
if not is_broken:
    print("Sensor is working")
````

## Part 4: Validating Data with Conditionals

This is where conditionals become useful for your project! Let's validate data:

````python
# Check if a single reading is valid
def is_valid_reading(value):
    """Check if a sensor reading is valid.
    
    Rules:
    - Must not be None
    - Must be a number
    - Must be in range 0-150
    """
    if value is None:
        return False
    
    if not isinstance(value, (int, float)):
        return False
    
    if value < 0 or value > 150:
        return False
    
    return True

# Test it
test_values = [25.0, -5, 200, None, "abc", 50, 0, 150]
for v in test_values:
    result = is_valid_reading(v)
    print(f"  {str(v):>6} → valid: {result}")
````

## Part 5: Building clean_data() with Rules

````python
def clean_data(data, config):
    """Clean data by applying validation rules.
    
    Args:
        data: list of dictionaries (rows)
        config: dict with cleaning settings
    
    Returns:
        list of clean dictionaries
    """
    cleaned = []
    dropped = 0
    
    threshold_min = config.get("min_value", 0)
    threshold_max = config.get("max_value", 100)
    
    for row in data:
        value = row.get("value")
        
        # Rule 1: skip if value is missing
        if value is None or value == "":
            dropped += 1
            continue    # skip to next row
        
        # Rule 2: try to convert to number
        try:
            num_value = float(value)
        except (ValueError, TypeError):
            dropped += 1
            continue
        
        # Rule 3: check range
        if num_value < threshold_min or num_value > threshold_max:
            dropped += 1
            continue
        
        # Passed all rules!
        row["value"] = num_value  # store as number
        cleaned.append(row)
    
    print(f"Cleaned: {len(data)} → {len(cleaned)} ({dropped} dropped)")
    return cleaned

# Test it
raw_data = [
    {"id": 1, "value": "25.0"},
    {"id": 2, "value": ""},        # missing
    {"id": 3, "value": "-10"},     # below min
    {"id": 4, "value": "abc"},     # not a number
    {"id": 5, "value": "50.5"},
    {"id": 6, "value": "200"},     # above max
    {"id": 7, "value": "75.0"},
]

config = {"min_value": 0, "max_value": 100}
result = clean_data(raw_data, config)

print("\nClean data:")
for row in result:
    print(f"  id={row['id']}, value={row['value']}")
````

**Expected Output:**
```
Cleaned: 7 → 3 (4 dropped)

Clean data:
  id=1, value=25.0
  id=5, value=50.5
  id=7, value=75.0
```
