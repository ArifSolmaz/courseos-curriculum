**AutoDashboard Reporter** works with IoT sensor time series (temperature, humidity, pressure).

Here's your sample data:

```python
# Sample data for IoT/Reporting track
sample_data = [{"ts": "2024-01-01 00:00", "device": "D01", "temp": 22.5, "humidity": 45},
 {"ts": "2024-01-01 01:00", "device": "D01", "temp": 23.0, "humidity": 47},
 {"ts": "2024-01-01 02:00", "device": "D02", "temp": "", "humidity": 50},
 {"ts": "2024-01-01 03:00", "device": "D01", "temp": 99.9, "humidity": 30},
 {"ts": "2024-01-01 04:00", "device": "D02", "temp": 21.8, "humidity": 48}]

print(f"Loaded {len(sample_data)} rows")
for row in sample_data:
    print(f"  {row}")
```

---
## Must-Pass Core

You MUST complete this before moving on.

```python
# Must-pass: Implement basic validation for temp
def clean_data(data, config):
    """Clean IoT/Reporting data."""
    cleaned = []
    dropped = 0
    
    for row in data:
        val = row.get("temp")
        
        # Rule: skip missing/invalid values
        if val is None or val == "":
            dropped += 1
            continue
        
        try:
            num = float(val)
        except (ValueError, TypeError):
            dropped += 1
            continue
        
        # Rule: must be in range (0, 60)
        low, high = (0, 60)
        if num < low or num > high:
            dropped += 1
            continue
        
        cleaned.append({**row, "temp": num})
    
    print(f"Cleaned: {len(data)} → {len(cleaned)} ({dropped} dropped)")
    return cleaned

# Test it
result = clean_data(sample_data, config)
print(f"\nClean rows:")
for row in result:
    print(f"  {row}")

assert len(result) < len(sample_data), "Should have dropped at least 1 row"
print("\nMust-pass core: PASSED!")
```
