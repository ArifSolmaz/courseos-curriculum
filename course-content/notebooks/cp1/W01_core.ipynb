## Part 3: What is a Pipeline?

A **pipeline** is a series of steps that transform data:

```
Raw Data → Clean → Analyze → Plot → Export
```

Think of it like a factory assembly line:
1. **Load** raw materials (data)
2. **Clean** — remove defective parts (bad data)
3. **Analyze** — measure and inspect
4. **Plot** — create visual reports
5. **Export** — package the results

This is what you will build over the next 14 weeks!

````python
# Here's a TINY preview of what your pipeline will look like:

# Step 1: Some raw data (just numbers for now)
raw_data = [10, 20, -5, 30, 999, 15]
print("Raw data:", raw_data)

# Step 2: Clean (remove negative and extreme values)
clean_data = []
for value in raw_data:
    if 0 <= value <= 100:
        clean_data.append(value)
print("Clean data:", clean_data)

# Step 3: Analyze (calculate average)
total = 0
for value in clean_data:
    total = total + value
average = total / len(clean_data)
print("Average:", average)

print("\nDon't worry if you don't understand everything yet!")
print("By Week 14, you will write all of this yourself.")
````
