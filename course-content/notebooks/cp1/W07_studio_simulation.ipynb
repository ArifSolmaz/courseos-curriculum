## Your Track Data

**SimLab Engine** works with simulation timestep logs with position, velocity, and score.

- **Primary value column:** `score`
- **Valid range:** 0 to 500 points
- **Alert threshold:** 25 points
- **Real-world use:** analyzing game simulation logs to find optimal strategies

Here is your sample dataset:

```python
sample_data = [
    {"step": 0, "x": 0.0, "y": 0.0, "velocity": 1.0, "score": 0},
    {"step": 1, "x": 1.0, "y": 0.5, "velocity": 1.2, "score": 10},
    {"step": 2, "x": 2.1, "y": 1.2, "velocity": -0.5, "score": 20},
    {"step": 3, "x": 3.0, "y": 2.0, "velocity": 1.5, "score": 35},
    {"step": 4, "x": 4.2, "y": 2.8, "velocity": 999, "score": -1},
    {"step": 5, "x": 5.0, "y": 3.5, "velocity": 1.8, "score": 50},
    {"step": 6, "x": 5.9, "y": 4.1, "velocity": 1.3, "score": 62},
    {"step": 7, "x": 6.5, "y": 4.9, "velocity": 0.0, "score": 70}
]

print(f"Loaded {len(sample_data)} rows")
print()
for i, row in enumerate(sample_data):
    print(f"  Row {i}: {row}")
```

### Step 1: Compute stats on score

```python
values = []
for row in sample_data:
    try:
        values.append(float(row["score"]))
    except (ValueError, TypeError, KeyError):
        pass

if values:
    total = sum(values)
    count = len(values)
    mean_val = total / count
    print(f"Count: {count}")
    print(f"Mean score: {mean_val:.2f} points")
    print(f"Min: {min(values)} points")
    print(f"Max: {max(values)} points")
    print()
    above = sum(1 for v in values if v > 25)
    print(f"Above threshold (25): {above}")
    print()
    print("Must-pass core: PASSED!")
else:
    print("ERROR: No valid values found!")
```
