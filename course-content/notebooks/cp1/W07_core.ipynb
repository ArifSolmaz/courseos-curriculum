## Part 1: List Indexing & Slicing

```python
data = [10, 20, 30, 40, 50, 60, 70, 80]

# Indexing (0-based)
print(f"First: {data[0]}")      # 10
print(f"Last: {data[-1]}")      # 80
print(f"Third: {data[2]}")      # 30

# Slicing [start:stop] (stop is excluded)
print(f"First 3: {data[0:3]}")  # [10, 20, 30]
print(f"Last 3: {data[-3:]}")   # [60, 70, 80]
print(f"Middle: {data[2:6]}")   # [30, 40, 50, 60]
```

---
## Part 2: Moving Window (Moving Average)

```python
def moving_average(values, window_size):
    """Calculate moving average with given window size.
    
    A moving average smooths data by averaging nearby values.
    For window_size=3: avg of [i-1, i, i+1]
    """
    result = []
    for i in range(len(values)):
        # Get the window
        start = max(0, i - window_size // 2)
        end = min(len(values), i + window_size // 2 + 1)
        window = values[start:end]
        avg = sum(window) / len(window)
        result.append(round(avg, 2))
    return result

data = [10, 50, 20, 40, 30, 60, 25]
smoothed = moving_average(data, 3)
print(f"Original:  {data}")
print(f"Smoothed:  {smoothed}")
```

## Part 3: List Operations

```python
numbers = [5, 2, 8, 1, 9, 3]

# Append and extend
numbers.append(10)
print(f"After append: {numbers}")

# Sort (modifies the list)
numbers.sort()
print(f"Sorted: {numbers}")

# Reverse
numbers.reverse()
print(f"Reversed: {numbers}")

# List comprehension (shortcut for creating lists)
squares = [x**2 for x in range(1, 6)]
print(f"Squares: {squares}")

# Filter with comprehension
data = [10, 25, 30, 15, 20, 35]
big = [x for x in data if x > 20]
print(f"Values > 20: {big}")
```
