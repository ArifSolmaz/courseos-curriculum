## Your Track Data

**CleanReport Pipeline** works with messy survey/CSV data with missing values and inconsistent types.

Here's your sample data:

```python
# Sample data for Data/AI track
sample_data = [{"id": 1, "age": "25", "score": "88.5", "city": "Cairo"},
 {"id": 2, "age": "", "score": "92.0", "city": "Alex"},
 {"id": 3, "age": "abc", "score": "75.0", "city": ""},
 {"id": 4, "age": "30", "score": "-10", "city": "Cairo"},
 {"id": 5, "age": "22", "score": "95.5", "city": "Luxor"}]

print(f"Loaded {len(sample_data)} rows")
for row in sample_data:
    print(f"  {row}")
```

---
## Must-Pass Core

You MUST complete this before moving on.

```python
# Must-pass: Implement basic validation for score
def clean_data(data, config):
    """Clean Data/AI data."""
    cleaned = []
    dropped = 0
    
    for row in data:
        val = row.get("score")
        
        # Rule: skip missing/invalid values
        if val is None or val == "":
            dropped += 1
            continue
        
        try:
            num = float(val)
        except (ValueError, TypeError):
            dropped += 1
            continue
        
        # Rule: must be in range (0, 100)
        low, high = (0, 100)
        if num < low or num > high:
            dropped += 1
            continue
        
        cleaned.append({**row, "score": num})
    
    print(f"Cleaned: {len(data)} → {len(cleaned)} ({dropped} dropped)")
    return cleaned

# Test it
result = clean_data(sample_data, config)
print(f"\nClean rows:")
for row in result:
    print(f"  {row}")

assert len(result) < len(sample_data), "Should have dropped at least 1 row"
print("\nMust-pass core: PASSED!")
```
