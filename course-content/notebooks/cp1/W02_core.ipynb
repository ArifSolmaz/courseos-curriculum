## Part 2: Data Types

Every value in Python has a **type**. The 4 basic types are:

````python
# int — whole numbers (no decimal point)
age = 20
print(type(age))      # <class 'int'>

# float — decimal numbers
temp = 36.6
print(type(temp))     # <class 'float'>

# str — text (must be in quotes)
name = "Alice"
print(type(name))     # <class 'str'>

# bool — True or False
is_valid = True
print(type(is_valid))  # <class 'bool'>
````

**Expected Output:**
```
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

### Converting Between Types

````python
# Converting types
x = "42"          # this is a string!
print(type(x))    # <class 'str'>

y = int(x)        # convert to integer
print(type(y))    # <class 'int'>
print(y + 10)     # now we can do math: 52

z = float("3.14") # string to float
print(z)          # 3.14

w = str(100)      # int to string
print("Value: " + w)  # "Value: 100"
````

## Part 3: Math Operations

````python
a = 15
b = 4

print(f"a + b = {a + b}")     # Addition: 19
print(f"a - b = {a - b}")     # Subtraction: 11
print(f"a * b = {a * b}")     # Multiplication: 60
print(f"a / b = {a / b}")     # Division: 3.75
print(f"a // b = {a // b}")   # Floor division: 3
print(f"a % b = {a % b}")     # Modulo (remainder): 3
print(f"a ** b = {a ** b}")   # Power: 50625
````

### Updating Variables

````python
count = 0
print(f"Start: {count}")

count = count + 1  # add 1
print(f"After +1: {count}")

count += 5         # shorthand for count = count + 5
print(f"After +=5: {count}")

count *= 2         # shorthand for count = count * 2
print(f"After *=2: {count}")
````

## Part 4: f-Strings (Formatted Text)

f-strings let you put variables inside text. Start the string with `f`:

````python
name = "Motor A"
rpm = 1500
temp = 72.456

# Basic f-string
print(f"Sensor: {name}")
print(f"RPM: {rpm}")

# Format decimals
print(f"Temperature: {temp:.1f} C")    # 1 decimal: 72.5
print(f"Temperature: {temp:.2f} C")    # 2 decimals: 72.46

# Math inside f-strings
print(f"RPM doubled: {rpm * 2}")
````

**Expected Output:**
```
Sensor: Motor A
RPM: 1500
Temperature: 72.5 C
Temperature: 72.46 C
RPM doubled: 3000
```

## Part 5: Your Config Dictionary

A **dictionary** stores key-value pairs. We use it for configuration:

````python
# A simple config dictionary
config = {
    "project_name": "my_project",
    "track": "robotics",
    "version": "v1",
    "threshold": 50.0,
    "data_points": 100,
}

# Access values using keys
print(f"Project: {config['project_name']}")
print(f"Track: {config['track']}")
print(f"Threshold: {config['threshold']}")
````

````python
# Adding new keys
config["author"] = "Student A"
print(config)

# Checking if a key exists
if "threshold" in config:
    print(f"Threshold is set to {config['threshold']}")
````

## Part 6: Using Config in a Function

````python
def load_data(config):
    """Load sample data based on config."""
    name = config["project_name"]
    n = config.get("data_points", 10)  # default to 10
    
    # Generate simple sample data
    data = []
    for i in range(n):
        data.append({"index": i, "value": i * 2.5 + 10})
    
    print(f"[{name}] Loaded {len(data)} rows")
    return data

# Use it!
config = {
    "project_name": "my_sensor",
    "data_points": 5,
}

result = load_data(config)
for row in result:
    print(f"  Row {row['index']}: value = {row['value']}")
````
