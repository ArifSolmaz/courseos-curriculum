```python
# Must-Pass: Implement validate_schema for Lightcurve Explorer
# Your schema should check for these columns: obs_id, timestamp, brightness, error_bar, filter

def validate_schema(data, config):
    """Validate Lightcurve Explorer data schema."""
    if not data:
        raise ValueError("Data is empty")
    
    required = config.get("required_columns", [])
    actual = set(data[0].keys())
    missing = set(required) - actual
    
    if missing:
        raise ValueError("Missing columns: " + str(sorted(missing)))
    
    print("Schema valid for Lightcurve Explorer")
    return True

config = {"required_columns": ['obs_id', 'timestamp', 'brightness', 'error_bar', 'filter']}

# TODO: Test with your sample data
# validate_schema(sample_data, config)
print("TODO: update sample_data to include all required columns and test")
```
