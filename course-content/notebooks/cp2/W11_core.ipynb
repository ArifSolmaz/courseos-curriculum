## Part 1: Why Modularize?

Right now, your functions are defined inside notebooks. This has problems:

- You cannot reuse functions across notebooks without copy-pasting
- If you fix a bug, you must fix it in every notebook
- Your notebooks are hundreds of lines long -- hard to read
- You cannot test functions independently

**Modularization** moves your functions into Python files (modules) that notebooks import from.

### Before vs. After Modularization

```
BEFORE (v1 style):               AFTER (v2 style):
+-------------------+            +------------------+
| notebook.ipynb    |            | notebook.ipynb   |
| - def load_data   |            | from src import  |
| - def clean_data  |            |   load_data,     |
| - def analyze     |            |   clean_data,    |
| - def plot        |            |   analyze, plot  |
| - def export      |            | # just calls     |
| (500+ lines)      |            | (50 lines)       |
+-------------------+            +------------------+
                                        |
                                 +------v----------+
                                 | src/            |
                                 |   __init__.py   |
                                 |   config.py     |
                                 |   io.py         |
                                 |   cleaning.py   |
                                 |   analysis.py   |
                                 |   plotting.py   |
                                 |   reporting.py  |
                                 +-----------------+
```

````python
# Your /src layout should look like this:
print("Target structure:")
print("  src/")
print("    project_name/")
print("      __init__.py      <- imports all public functions")
print("      config.py        <- get_config()")
print("      io.py            <- load_data()")
print("      cleaning.py      <- clean_data(), validate_schema()")
print("      analysis.py      <- analyze()")
print("      plotting.py      <- plot()")
print("      reporting.py     <- export_results(), self_check()")
print()
print("Notebooks should IMPORT from /src, not define functions!")
````

### Step-by-Step Migration

````python
# Step 1: Create __init__.py that imports from modules
init_code = """
from .config import get_config
from .io import load_data
from .cleaning import clean_data, validate_schema
from .analysis import analyze
from .plotting import plot
from .reporting import export_results, self_check
"""
print("Example __init__.py:")
print(init_code)

# Step 2: Notebook cell after migration
notebook_code = """
import sys
sys.path.insert(0, "src")
from project_name import get_config, load_data, clean_data, analyze, plot, export_results

config = get_config()
data = load_data(config)
cleaned = clean_data(data, config)
results = analyze(cleaned, config)
figures = plot(cleaned, results, config)
exported = export_results(cleaned, results, figures, config)
"""
print("Example notebook usage:")
print(notebook_code)
````

## Part 3: Example Module Files

Here is what each module file should look like:

````python
# Example: config.py
config_py_example = """
import os

def get_config(base_dir=None):
    if base_dir is None:
        for candidate in [".", "..", "../.."]:
            if os.path.exists(os.path.join(candidate, "src")):
                base_dir = candidate
                break
        else:
            base_dir = "."
    return {
        "project_name": "my_project",
        "track": "data",
        "version": "v2",
        "raw_data_path": os.path.join(base_dir, "data/raw/sample.csv"),
        "required_columns": ["id", "value"],
        "numeric_columns": ["value"],
        "value_ranges": {"value": (0, 100)},
    }
"""
print("=== config.py ===")
print(config_py_example)
````

````python
# Example: cleaning.py
cleaning_py_example = """
def validate_schema(data, config):
    if not data:
        raise ValueError("Data is empty")
    required = config.get("required_columns", [])
    actual = set(data[0].keys())
    missing = set(required) - actual
    if missing:
        raise ValueError("Missing columns: " + str(sorted(missing)))
    return True

def clean_data(data, config):
    cleaned = []
    for row in data:
        try:
            for col in config.get("numeric_columns", []):
                val = float(row.get(col, ""))
                ranges = config.get("value_ranges", {})
                if col in ranges:
                    low, high = ranges[col]
                    if val < low or val > high:
                        break
                row[col] = val
            else:
                cleaned.append(row)
        except (ValueError, TypeError):
            pass
    return cleaned
"""
print("=== cleaning.py ===")
print(cleaning_py_example)
````

## Part 4: Testing Your Modules

After migrating, verify everything still works.

````python
# Test: simulate importing from modules
# (In real code: from project_name import get_config, load_data, ...)

def test_module_imports():
    """Test that all module functions work after migration."""
    print("=== Module Import Test ===\n")
    
    # Test config
    try:
        import os
        config = {"project_name": "test", "required_columns": ["id", "value"],
                  "numeric_columns": ["value"], "value_ranges": {"value": (0, 100)}}
        assert "project_name" in config
        print("[PASS] get_config works")
    except Exception as e:
        print("[FAIL] get_config: " + str(e))
    
    # Test validation
    try:
        data = [{"id": 1, "value": "25"}]
        actual = set(data[0].keys())
        required = set(config.get("required_columns", []))
        assert required.issubset(actual)
        print("[PASS] validate_schema works")
    except Exception as e:
        print("[FAIL] validate_schema: " + str(e))
    
    # Test cleaning
    try:
        test_data = [{"id": 1, "value": "25"}, {"id": 2, "value": "abc"}]
        cleaned = []
        for row in test_data:
            try:
                float(row["value"])
                cleaned.append(row)
            except ValueError:
                pass
        assert len(cleaned) == 1
        print("[PASS] clean_data works")
    except Exception as e:
        print("[FAIL] clean_data: " + str(e))
    
    print("\nAll module tests passed!")

test_module_imports()
````

### Common Mistakes: Module Migration

**Mistake 1:** Forgetting __init__.py -- Python will not recognize the directory as a package.

**Fix:** Always create __init__.py, even if it just contains imports.

**Mistake 2:** Circular imports -- module A imports from B, and B imports from A.

**Fix:** Design modules so dependencies flow one way: config -> io -> cleaning -> analysis.

**Mistake 3:** Not updating sys.path -- notebooks cannot find /src.

**Fix:** Add sys.path.insert(0, 'src') at the top of every notebook.


### Key Takeaway

- Move functions from notebooks to .py files in /src
- Use __init__.py to control what gets imported
- Notebooks become short: import + call + display
- One function, one place -- fix a bug once, fixed everywhere
- Test all imports after migration to catch problems early
