## Part 1: NumPy Arrays

A NumPy array is like a Python list, but optimized for math. When you do math on a NumPy array, it operates on ALL elements at once -- no loop needed. This is called **vectorization**.

### List vs. NumPy Array

```
Python List (slow):            NumPy Array (fast):
  [1, 2, 3] * 2                [1, 2, 3] * 2
  = [1, 2, 3, 1, 2, 3]         = [2, 4, 6]
  (repeats the list!)           (multiplies each element!)
```

````python
import numpy as np

# Creating arrays
a = np.array([1, 2, 3, 4, 5])
print("Array:", a)
print("Type:", type(a))
print("Shape:", a.shape)
print("Dtype:", a.dtype)
print()

# Vectorized operations -- NO LOOPS!
print("a * 2 =", a * 2)
print("a + 10 =", a + 10)
print("a ** 2 =", a ** 2)
print()

# Statistical functions
print("Mean:", a.mean())
print("Std:", a.std())
print("Sum:", a.sum())
print("Min:", a.min())
print("Max:", a.max())
````

```

```

### Replacing Loops With NumPy

````python
# BEFORE: loop version (slow for large data)
data_list = list(range(1000))
result_loop = []
for x in data_list:
    result_loop.append(x * 2 + 5)

# AFTER: NumPy version (fast, one line)
data_np = np.arange(1000)
result_np = data_np * 2 + 5

print("Loop result[:5]:", result_loop[:5])
print("NumPy result[:5]:", result_np[:5].tolist())
print("Same?", result_loop[:5] == result_np[:5].tolist())
````

**Expected Output:**
```
Loop result[:5]: [5, 7, 9, 11, 13]
NumPy result[:5]: [5, 7, 9, 11, 13]
Same? True
```

## Part 2: NumPy for Data Analysis

````python
# Statistical analysis with NumPy
data = np.array([22.5, 18.3, 30.1, 25.7, 19.8, 27.4, 23.1, 28.6])

print("Descriptive Statistics:")
print("  Mean:", round(np.mean(data), 2))
print("  Median:", round(np.median(data), 2))
print("  Std:", round(np.std(data), 2))
print("  Min:", round(np.min(data), 2))
print("  Max:", round(np.max(data), 2))
print()

# Boolean indexing (vectorized filtering)
mask = data > 25
print("Values > 25:", data[mask])
print("Count > 25:", np.sum(mask))
print("Percent > 25:", round(np.mean(mask) * 100, 1), "%")
````

**Expected Output:**
```
Descriptive Statistics:
  Mean: 24.44
  Median: 24.4
  Std: 3.95
  Min: 18.3
  Max: 30.1

Values > 25: [30.1 25.7 27.4 28.6]
Count > 25: 4
Percent > 25: 50.0 %
```

## Part 3: Speed Comparison

````python
import time

n = 100_000

# Loop version
data_list = list(range(n))
start = time.time()
result_loop = [x ** 2 + 2 * x + 1 for x in data_list]
loop_time = time.time() - start

# NumPy version
data_np = np.arange(n)
start = time.time()
result_np = data_np ** 2 + 2 * data_np + 1
np_time = time.time() - start

print("Loop:  " + str(round(loop_time * 1000, 2)) + " ms")
print("NumPy: " + str(round(np_time * 1000, 2)) + " ms")
speedup = loop_time / np_time if np_time > 0 else 0
print("NumPy is ~" + str(round(speedup, 1)) + "x faster!")
````

**Expected Output:**
```
Loop:  ~30 ms
NumPy: ~1 ms
NumPy is ~30x faster!
(actual times depend on your machine)
```

## Part 4: NumPy in Your Pipeline

Let us see how to use NumPy inside your actual pipeline functions.

````python
import numpy as np

def analyze_with_numpy(clean_data, value_col="value"):
    """Analyze data using NumPy for speed and convenience.
    
    Args:
        clean_data: list of dicts with numeric values
        value_col: which column to analyze
    
    Returns:
        dict with analysis_summary
    """
    # Extract values into a NumPy array
    values = np.array([row[value_col] for row in clean_data])
    
    summary = {
        "count": len(values),
        "mean": round(float(np.mean(values)), 2),
        "std": round(float(np.std(values)), 2),
        "median": round(float(np.median(values)), 2),
        "min": round(float(np.min(values)), 2),
        "max": round(float(np.max(values)), 2),
        "q25": round(float(np.percentile(values, 25)), 2),
        "q75": round(float(np.percentile(values, 75)), 2),
    }
    
    print("Analysis (NumPy):")
    for k, v in summary.items():
        print("  " + k + ": " + str(v))
    
    return {"analysis_summary": summary}

# Test with sample data
import random
random.seed(42)
clean_data = [{"id": i, "value": round(random.gauss(50, 15), 2)} for i in range(100)]
results = analyze_with_numpy(clean_data)
````

**Expected Output:**
```

  std: ~15
  median: ~50
  min: ~10
  max: ~90
  q25: ~40
  q75: ~60
(approximate)
```

### NumPy Filtering vs. Loop Filtering

````python
# Compare filtering approaches
values = np.array([row["value"] for row in clean_data])

# NumPy way: boolean indexing
mask = (values >= 30) & (values <= 70)  # both conditions at once
filtered_np = values[mask]
print("NumPy filtered: " + str(len(filtered_np)) + " values in [30, 70]")
print("  Mean of filtered:", round(float(np.mean(filtered_np)), 2))

# How many are outliers? (beyond 2 std from mean)
mean_v = np.mean(values)
std_v = np.std(values)
outlier_mask = np.abs(values - mean_v) > 2 * std_v
print("\nOutliers (>2 std):", int(np.sum(outlier_mask)))
if np.any(outlier_mask):
    print("  Values:", values[outlier_mask].round(2).tolist())
````

### Common Mistakes: NumPy

**Mistake 1:** Mixing Python lists and NumPy arrays -- np.array([1,2]) + [3,4] works but may surprise you.

**Fix:** Convert to NumPy early and stay in NumPy for all calculations.

**Mistake 2:** Using Python sum() instead of np.sum() -- Python sum is slower on arrays.

**Fix:** Always use np.sum, np.mean, etc. for NumPy arrays.

**Mistake 3:** Forgetting to convert back to Python float for JSON -- np.float64 is not JSON-serializable.

**Fix:** Use float(np_value) before putting NumPy results into JSON reports.


### Key Takeaway

- NumPy arrays do math on ALL elements at once (vectorization)
- No loops needed for element-wise operations
- NumPy is 10-100x faster than Python loops for numeric work
- Use np.mean, np.std, np.median for statistics
- Boolean indexing (arr[mask]) replaces filter loops
- Convert np.float64 to float() before JSON serialization
