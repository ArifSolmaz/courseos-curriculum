## Part 1: v2 Release Checklist

Your v2 pipeline must pass ALL of these checks before demo.

````python
import os, json

def v2_release_check():
    """Final v2 release validation."""
    print("=== v2 Release Check ===\n")
    
    passed = 0
    failed = 0
    
    # Check 1: Required files
    required_files = [
        "data/cleaned/cleaned.csv",
        "reports/report.json",
        "reports/figures/timeseries.png",
        "reports/figures/summary.png",
    ]
    
    print("--- File Checks ---")
    for f in required_files:
        if os.path.exists(f) and os.path.getsize(f) > 0:
            size = os.path.getsize(f)
            print("  [PASS] " + f + " (" + str(size) + " bytes)")
            passed += 1
        else:
            print("  [FAIL] " + f + " missing or empty")
            failed += 1
    
    # Check 2: Report schema
    print("\n--- Report Checks ---")
    required_keys = [
        "project_name", "track", "version",
        "dataset", "cleaning_summary", "analysis_summary",
    ]
    try:
        with open("reports/report.json") as f:
            report = json.load(f)
        for key in required_keys:
            if key in report:
                print("  [PASS] report has " + repr(key))
                passed += 1
            else:
                print("  [FAIL] report missing " + repr(key))
                failed += 1
    except FileNotFoundError:
        print("  [FAIL] report.json not found")
        failed += len(required_keys)
    except json.JSONDecodeError:
        print("  [FAIL] report.json is not valid JSON")
        failed += len(required_keys)
    
    # Check 3: Analysis metrics
    print("\n--- Analysis Checks ---")
    try:
        summary = report.get("analysis_summary", {})
        n_metrics = len([v for v in summary.values() if isinstance(v, (int, float))])
        if n_metrics >= 3:
            print("  [PASS] " + str(n_metrics) + " numeric metrics")
            passed += 1
        else:
            print("  [FAIL] Only " + str(n_metrics) + " metrics (need 3+)")
            failed += 1
    except Exception:
        print("  [FAIL] Could not check metrics")
        failed += 1
    
    print("\n=== " + str(passed) + " passed, " + str(failed) + " failed ===")
    if failed == 0:
        print("\nv2 is READY for demo! Congratulations!")
    else:
        print("\nFix " + str(failed) + " failure(s) before demo.")
    return failed == 0

v2_release_check()
````

## Part 4: Demo Script

Write a script that runs your entire pipeline for the demo.

````python
def run_demo():
    """Complete v2 pipeline demo."""
    import random, os
    random.seed(42)
    
    print("=" * 50)
    print("  v2 PIPELINE DEMO")
    print("=" * 50)
    
    # Step 1: Load
    print("\n--- Step 1: Load Data ---")
    data = [{"id": i, "value": str(round(random.gauss(50, 20), 2))}
            for i in range(100)]
    data.append({"id": 100, "value": ""})
    data.append({"id": 101, "value": "abc"})
    print("Loaded " + str(len(data)) + " rows")
    
    # Step 2: Validate
    print("\n--- Step 2: Validate Schema ---")
    actual = set(data[0].keys())
    required = {"id", "value"}
    missing = required - actual
    if not missing:
        print("Schema valid: " + str(len(required)) + " required columns present")
    else:
        print("FAILED: missing " + str(missing))
    
    # Step 3: Clean
    print("\n--- Step 3: Clean Data ---")
    cleaned = []
    drops = {}
    for row in data:
        try:
            v = float(row["value"])
            if 0 <= v <= 100:
                cleaned.append({**row, "value": v})
            else:
                reason = "out_of_range"
                drops[reason] = drops.get(reason, 0) + 1
        except (ValueError, TypeError):
            val = row.get("value", "")
            reason = "missing" if not val or str(val).strip() == "" else "non_numeric"
            drops[reason] = drops.get(reason, 0) + 1
    print("Cleaned: " + str(len(data)) + " -> " + str(len(cleaned)) + " rows")
    for reason, count in drops.items():
        print("  Dropped (" + reason + "): " + str(count))
    
    # Step 4: Analyze
    print("\n--- Step 4: Analyze ---")
    values = [r["value"] for r in cleaned]
    mean_v = sum(values) / len(values)
    std_v = (sum((x - mean_v)**2 for x in values) / len(values))**0.5
    print("  count: " + str(len(values)))
    print("  mean:  " + str(round(mean_v, 2)))
    print("  std:   " + str(round(std_v, 2)))
    print("  min:   " + str(round(min(values), 2)))
    print("  max:   " + str(round(max(values), 2)))
    
    # Step 5: Export
    print("\n--- Step 5: Export ---")
    os.makedirs("reports", exist_ok=True)
    import json
    report = {
        "project_name": "demo",
        "track": "data",
        "version": "v2",
        "dataset": {"n_raw": len(data), "n_clean": len(cleaned)},
        "analysis_summary": {"mean": round(mean_v, 2), "std": round(std_v, 2),
                             "min": round(min(values), 2), "max": round(max(values), 2)},
        "cleaning_summary": drops,
    }
    with open("reports/report.json", "w") as f:
        json.dump(report, f, indent=2)
    print("Saved: reports/report.json")
    
    print("\n" + "=" * 50)
    print("  v2 DEMO COMPLETE")
    print("=" * 50)

run_demo()
````
