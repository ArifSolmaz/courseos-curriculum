## Part 1: Expanding self_check()

Your v1 self_check probably just tested "does it run?" Your v2 self_check must test:

1. **Happy path** -- normal inputs work correctly
2. **Edge cases** -- empty data, single row, all same values
3. **Error handling** -- bad inputs give clear errors
4. **Output format** -- results have the right structure

````python
def self_check():
    """Comprehensive self-check for v2 pipeline."""
    print("=== v2 Self-Check ===\n")
    passed = 0
    failed = 0
    
    # --- Happy Path Tests ---
    print("--- Happy Path ---")
    
    # Test 1: Clean data returns list
    try:
        test_data = [{"value": "10"}, {"value": "20"}, {"value": "30"}]
        result = [r for r in test_data if r["value"] != ""]  # simulate clean
        assert isinstance(result, list)
        assert len(result) == 3
        print("[PASS] clean_data returns correct list")
        passed += 1
    except Exception as e:
        print("[FAIL] clean_data: " + str(e))
        failed += 1
    
    # Test 2: Missing values dropped
    try:
        test_data = [{"value": "10"}, {"value": ""}, {"value": "20"}]
        result = [r for r in test_data if r["value"] != ""]
        assert len(result) == 2, "Expected 2, got " + str(len(result))
        print("[PASS] Missing values dropped")
        passed += 1
    except Exception as e:
        print("[FAIL] Missing values: " + str(e))
        failed += 1
    
    # --- Edge Cases ---
    print("\n--- Edge Cases ---")
    
    # Test 3: Empty input
    try:
        result = []
        assert isinstance(result, list)
        assert len(result) == 0
        print("[PASS] Handles empty input")
        passed += 1
    except Exception as e:
        print("[FAIL] Empty input: " + str(e))
        failed += 1
    
    # Test 4: Single row
    try:
        test_data = [{"value": "42"}]
        result = [r for r in test_data if r["value"] != ""]
        assert len(result) == 1
        print("[PASS] Handles single row")
        passed += 1
    except Exception as e:
        print("[FAIL] Single row: " + str(e))
        failed += 1
    
    # Test 5: All same values
    try:
        values = [42.0] * 10
        mean_v = sum(values) / len(values)
        assert mean_v == 42.0
        print("[PASS] Handles uniform data")
        passed += 1
    except Exception as e:
        print("[FAIL] Uniform data: " + str(e))
        failed += 1
    
    # --- Output Format ---
    print("\n--- Output Format ---")
    
    # Test 6: Results have required keys
    try:
        results = {"analysis_summary": {"mean": 15, "std": 5, "count": 2}}
        assert "analysis_summary" in results
        assert len(results["analysis_summary"]) >= 3
        print("[PASS] Results have required keys")
        passed += 1
    except Exception as e:
        print("[FAIL] Results format: " + str(e))
        failed += 1
    
    print("\n=== " + str(passed) + " passed, " + str(failed) + " failed ===")
    return failed == 0

self_check()
````

```

[PASS] Handles single row
[PASS] Handles uniform data

[PASS] Results have required keys
```

## Part 2: The Arrange-Act-Assert Pattern

Every test should follow this pattern:

1. **Arrange** -- set up test data
2. **Act** -- call the function
3. **Assert** -- check the result

````python
# Example: well-structured test
def test_type_conversion():
    """Test that numeric strings are converted to floats."""
    # ARRANGE: set up test data
    test_data = [{"value": "25.5"}, {"value": "30.0"}]
    
    # ACT: run the function
    result = []
    for row in test_data:
        result.append({"value": float(row["value"])})
    
    # ASSERT: check the result
    assert result[0]["value"] == 25.5
    assert result[1]["value"] == 30.0
    assert isinstance(result[0]["value"], float)
    print("[PASS] Type conversion works")

test_type_conversion()
````

**Expected Output:**
```
[PASS] Type conversion works
```

## Part 3: Testing Error Handling

Your tests should verify that bad inputs produce clear errors, not crashes.

````python
def test_error_handling():
    """Test that pipeline gives clear errors for bad input."""
    print("--- Error Handling Tests ---")
    passed = 0
    failed = 0
    
    # Test: None input gives clear error
    try:
        # Your validate_schema should handle this
        data = None
        if data is None:
            raise ValueError("Data cannot be None")
        print("[FAIL] Should have raised ValueError")
        failed += 1
    except ValueError as e:
        assert "None" in str(e) or "empty" in str(e).lower() or "cannot" in str(e).lower()
        print("[PASS] None input gives clear error")
        passed += 1
    except Exception as e:
        print("[FAIL] Wrong error type: " + type(e).__name__)
        failed += 1
    
    # Test: wrong type gives clear error
    try:
        data = "not a list"
        if not isinstance(data, list):
            raise TypeError("Expected list, got " + type(data).__name__)
        print("[FAIL] Should have raised TypeError")
        failed += 1
    except TypeError as e:
        print("[PASS] Wrong type gives clear error")
        passed += 1
    
    # Test: missing config key gives clear error
    try:
        config = {}  # missing required keys
        required_key = "required_columns"
        if required_key not in config:
            raise KeyError("Config missing required key: " + required_key)
        failed += 1
    except KeyError as e:
        print("[PASS] Missing config key gives clear error")
        passed += 1
    
    print("\n" + str(passed) + " passed, " + str(failed) + " failed")
    return failed == 0

test_error_handling()
````

**Expected Output:**
```
--- Error Handling Tests ---
[PASS] None input gives clear error
[PASS] Wrong type gives clear error
[PASS] Missing config key gives clear error

3 passed, 0 failed
```

## Part 4: Organizing Tests

Group your tests into categories for clarity.

````python
def run_all_tests():
    """Run all test categories and print a summary."""
    print("=" * 50)
    print("  v2 PIPELINE TEST SUITE")
    print("=" * 50)
    
    all_passed = 0
    all_failed = 0
    
    # Category 1: Happy path
    print("\n[Category 1: Happy Path]")
    tests = [
        ("data loads", True),
        ("cleaning works", True),
        ("analysis produces results", True),
    ]
    for name, result in tests:
        status = "[PASS]" if result else "[FAIL]"
        print("  " + status + " " + name)
        if result:
            all_passed += 1
        else:
            all_failed += 1
    
    # Category 2: Edge cases
    print("\n[Category 2: Edge Cases]")
    tests = [
        ("empty input", True),
        ("single row", True),
        ("all same values", True),
        ("all missing values", True),
    ]
    for name, result in tests:
        status = "[PASS]" if result else "[FAIL]"
        print("  " + status + " " + name)
        if result:
            all_passed += 1
        else:
            all_failed += 1
    
    # Category 3: Output format
    print("\n[Category 3: Output Format]")
    tests = [
        ("results has analysis_summary", True),
        ("3+ numeric metrics", True),
    ]
    for name, result in tests:
        status = "[PASS]" if result else "[FAIL]"
        print("  " + status + " " + name)
        if result:
            all_passed += 1
        else:
            all_failed += 1
    
    print("\n" + "=" * 50)
    print("  TOTAL: " + str(all_passed) + " passed, "
          + str(all_failed) + " failed")
    print("=" * 50)
    
    if all_failed == 0:
        print("  ALL TESTS PASSED!")
    return all_failed == 0

run_all_tests()
````

**Expected Output:**
```
==================================================
  v2 PIPELINE TEST SUITE
==================================================

[Category 1: Happy Path]
  [PASS] data loads
  [PASS] cleaning works
  [PASS] analysis produces results

[Category 2: Edge Cases]
  [PASS] empty input
  [PASS] single row
  [PASS] all same values
  [PASS] all missing values

[Category 3: Output Format]
  [PASS] results has analysis_summary
  [PASS] 3+ numeric metrics

==================================================
  TOTAL: 9 passed, 0 failed
==================================================
  ALL TESTS PASSED!
```

### Debugging Tips: Testing

- If a test fails, read the assertion message -- it tells you what went wrong
- Print the actual vs expected values when debugging failing tests
- Run tests one at a time to isolate failures
- A test that never fails is not testing anything useful
- If you fix a bug, write a test for it BEFORE fixing it

### Key Takeaway

- Test happy path, edge cases, and error handling
- Use Arrange-Act-Assert pattern for clear test structure
- Name your tests clearly (test_what_when_then)
- Run self_check() after EVERY change to catch regressions
- Organize tests into categories for readability
