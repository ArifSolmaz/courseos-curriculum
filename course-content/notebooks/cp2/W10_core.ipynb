## Part 1: What Are Golden Outputs?

A **golden output** is a known-correct result for a known input. You create a small, hand-crafted dataset where you can calculate the correct answer by hand. Then you run your pipeline and verify it matches.

If you change your code and the golden test fails, you know something broke.

### Golden Output Testing

```
GOLDEN DATASET (fixed, small)     EXPECTED RESULTS (calculated by hand)
+-----+-------+----------+       +-------------------+
| id  | value | category |       | n_clean: 5        |
+-----+-------+----------+       | mean: 30.0        |
| 1   | 10.0  | A        |       | min: 10.0         |
| 2   | 20.0  | A        |       | max: 50.0         |
| 3   | 30.0  | B        |       +-------------------+
| 4   |       | B        | <-- dropped (missing)
| 5   | abc   | A        | <-- dropped (non-numeric)
| 6   | 40.0  | B        |
| 7   | 50.0  | A        |
| 8   | -10   | B        | <-- dropped (out of range)
+-----+-------+----------+
```

````python
import json

def create_golden_dataset():
    """Create a deterministic toy dataset for testing.
    
    This dataset has known properties:
    - 8 rows total
    - 3 should be dropped (1 missing, 1 non-numeric, 1 out of range)
    - 5 clean rows with values: 10, 20, 30, 40, 50
    - Mean = 30.0, Min = 10.0, Max = 50.0
    """
    return [
        {"id": 1, "value": "10.0", "category": "A"},
        {"id": 2, "value": "20.0", "category": "A"},
        {"id": 3, "value": "30.0", "category": "B"},
        {"id": 4, "value": "",     "category": "B"},   # missing
        {"id": 5, "value": "abc",  "category": "A"},   # non-numeric
        {"id": 6, "value": "40.0", "category": "B"},
        {"id": 7, "value": "50.0", "category": "A"},
        {"id": 8, "value": "-10",  "category": "B"},   # out of range
    ]

GOLDEN_EXPECTED = {
    "n_raw": 8,
    "n_clean": 5,
    "n_dropped": 3,
    "mean": 30.0,
    "min": 10.0,
    "max": 50.0,
}

def test_golden():
    """Test pipeline against known expected results."""
    print("=== Golden Output Test ===\n")
    data = create_golden_dataset()
    assert len(data) == GOLDEN_EXPECTED["n_raw"]
    
    # Clean (simulate your pipeline)
    clean = []
    for row in data:
        try:
            val = float(row["value"])
            if val >= 0:
                clean.append({**row, "value": val})
        except (ValueError, TypeError):
            pass
    
    # Verify
    assert len(clean) == GOLDEN_EXPECTED["n_clean"], \
        "Expected " + str(GOLDEN_EXPECTED["n_clean"]) + " clean rows, got " + str(len(clean))
    
    values = [r["value"] for r in clean]
    actual_mean = sum(values) / len(values)
    assert abs(actual_mean - GOLDEN_EXPECTED["mean"]) < 1e-6, \
        "Mean: expected " + str(GOLDEN_EXPECTED["mean"]) + ", got " + str(actual_mean)
    
    assert min(values) == GOLDEN_EXPECTED["min"]
    assert max(values) == GOLDEN_EXPECTED["max"]
    
    print("[PASS] n_clean = " + str(len(clean)))
    print("[PASS] mean = " + str(actual_mean))
    print("[PASS] min = " + str(min(values)))
    print("[PASS] max = " + str(max(values)))
    print("\nGolden test PASSED!")

test_golden()
````

**Expected Output:**
```
=== Golden Output Test ===

[PASS] n_clean = 5
[PASS] mean = 30.0
[PASS] min = 10.0
[PASS] max = 50.0

Golden test PASSED!
```

## Part 2: Golden Tests for Groups

Test that group-level results are also correct.

````python
def test_golden_groups():
    """Test group-level results against golden expected."""
    print("=== Golden Group Test ===\n")
    data = create_golden_dataset()
    
    # Clean
    clean = []
    for row in data:
        try:
            val = float(row["value"])
            if val >= 0:
                clean.append({**row, "value": val})
        except (ValueError, TypeError):
            pass
    
    # Group by category
    groups = {}
    for row in clean:
        cat = row["category"]
        if cat not in groups:
            groups[cat] = []
        groups[cat].append(row["value"])
    
    # Expected group results (calculated by hand)
    # A: [10, 20, 50] -> mean = 26.67
    # B: [30, 40] -> mean = 35.0
    expected_groups = {
        "A": {"count": 3, "mean": 26.67},
        "B": {"count": 2, "mean": 35.0},
    }
    
    for cat, expected in expected_groups.items():
        actual_count = len(groups.get(cat, []))
        actual_vals = groups.get(cat, [])
        actual_mean = sum(actual_vals) / len(actual_vals) if actual_vals else 0
        
        assert actual_count == expected["count"], \
            "Group " + cat + " count: expected " + str(expected["count"]) + ", got " + str(actual_count)
        assert abs(actual_mean - expected["mean"]) < 0.1, \
            "Group " + cat + " mean: expected " + str(expected["mean"]) + ", got " + str(round(actual_mean, 2))
        
        print("[PASS] Group " + cat + ": count=" + str(actual_count)
              + ", mean=" + str(round(actual_mean, 2)))
    
    print("\nGolden group test PASSED!")

test_golden_groups()
````

**Expected Output:**
```
=== Golden Group Test ===

[PASS] Group A: count=3, mean=26.67
[PASS] Group B: count=2, mean=35.0

Golden group test PASSED!
```

## Part 3: Saving Golden Expected Results

Save your expected results to a JSON file so they are version-controlled.

````python
import json, os

def save_golden_expected(expected, filepath="tests/golden_expected.json"):
    """Save golden expected results to JSON."""
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    with open(filepath, "w") as f:
        json.dump(expected, f, indent=2)
    print("Saved golden expected: " + filepath)

def load_golden_expected(filepath="tests/golden_expected.json"):
    """Load golden expected from JSON."""
    with open(filepath) as f:
        return json.load(f)

# Save
save_golden_expected(GOLDEN_EXPECTED)

# Load and verify
loaded = load_golden_expected()
print("Loaded:", loaded)
assert loaded == GOLDEN_EXPECTED
print("Verified: loaded matches original!")
````

**Expected Output:**
```
Saved golden expected: tests/golden_expected.json
Loaded: {'n_raw': 8, 'n_clean': 5, 'n_dropped': 3, 'mean': 30.0, 'min': 10.0, 'max': 50.0}
Verified: loaded matches original!
```

## Part 4: Regression Testing

A **regression** is when code that used to work stops working. Golden tests catch regressions automatically.

````python
def regression_test():
    """Run full regression test suite."""
    print("=== Regression Test Suite ===\n")
    all_pass = True
    
    # Test 1: Golden dataset
    try:
        test_golden()
    except AssertionError as e:
        print("[REGRESSION] Golden test failed: " + str(e))
        all_pass = False
    
    # Test 2: Golden groups
    try:
        test_golden_groups()
    except AssertionError as e:
        print("[REGRESSION] Group test failed: " + str(e))
        all_pass = False
    
    # Test 3: Output format
    try:
        data = create_golden_dataset()
        clean = []
        for row in data:
            try:
                val = float(row["value"])
                if val >= 0:
                    clean.append({**row, "value": val})
            except (ValueError, TypeError):
                pass
        assert len(clean) > 0, "No clean data"
        assert all("value" in r for r in clean), "Missing value column"
        assert all(isinstance(r["value"], float) for r in clean), "Values not converted"
        print("[PASS] Output format correct")
    except AssertionError as e:
        print("[REGRESSION] Format test failed: " + str(e))
        all_pass = False
    
    if all_pass:
        print("\n=== All regression tests PASSED ===")
    else:
        print("\n=== REGRESSION DETECTED -- fix before committing ===")
    return all_pass

regression_test()
````

**Expected Output:**
```
=== Regression Test Suite ===

=== Golden Output Test ===
[PASS] n_clean = 5
[PASS] mean = 30.0
[PASS] min = 10.0
[PASS] max = 50.0
Golden test PASSED!

=== Golden Group Test ===
[PASS] Group A: count=3, mean=26.67
[PASS] Group B: count=2, mean=35.0
Golden group test PASSED!

[PASS] Output format correct

=== All regression tests PASSED ===
```

### Key Takeaway

- Golden datasets are small, hand-crafted, with known correct results
- Run golden tests after EVERY code change to catch regressions
- Include edge cases in golden data (missing, non-numeric, out of range)
- Golden tests document what your pipeline SHOULD do
- Save expected results to JSON for version control
