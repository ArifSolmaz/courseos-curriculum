## Your Track Data

**Lightcurve Explorer** works with star brightness (flux) measurements over time for exoplanet detection.

Here is your sample dataset:

```python
sample_data = [
    {'time': 0.0, 'flux': 1.0, 'sector': 'A'},
    {'time': 0.1, 'flux': 0.998, 'sector': 'A'},
    {'time': 0.2, 'flux': 0.7, 'sector': 'A'},
    {'time': 0.3, 'flux': 0.985, 'sector': 'A'},
    {'time': 0.4, 'flux': 1.01, 'sector': 'B'},
    {'time': 0.5, 'flux': 0.68, 'sector': 'B'},
    {'time': 0.6, 'flux': 0.995, 'sector': 'B'},
    {'time': 0.7, 'flux': 0.95, 'sector': 'B'}
]

print("Loaded " + str(len(sample_data)) + " rows")
print()
for i, row in enumerate(sample_data):
    print("  Row " + str(i) + ": " + str(row))
```

### Step 1: Implement binary search on your data

```python
import bisect

# Sort your data by the primary value column
sorted_data = sorted(sample_data, key=lambda r: r["flux"])
values = [r["flux"] for r in sorted_data]

print("Sorted values: " + str(values))
print()

# Use bisect to find values in a range
threshold = 0.9
idx = bisect.bisect_left(values, threshold)
print("Values >= " + str(threshold) + ":")
for row in sorted_data[idx:]:
    print("  " + str(row))
```
