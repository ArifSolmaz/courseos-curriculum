## Section 2: Your First Class

A class is created with the `class` keyword. The special method `__init__` is called automatically whenever you create a new object. The parameter `self` refers to the specific object being created or used -- it is how the object talks about itself.

Let us build a `Sensor` class step by step.

### Example 1: The Simplest Possible Class

````python
class Sensor:
    """A sensor with a name."""
    pass  # empty class -- valid but useless

s = Sensor()
print(type(s))
print(s)
````

### Example 2: Adding `__init__` (the Constructor)

`__init__` runs automatically when you create an object. It sets up the object's initial state (its attributes). Think of it as filling out the 'birth certificate' for the object.

````python
class Sensor:
    """A sensor with a name and unit."""

    def __init__(self, name, unit):
        self.name = name    # attribute: the sensor's name
        self.unit = unit    # attribute: what it measures

# Create two sensor objects
temp = Sensor("Temperature", "Celsius")
rpm = Sensor("RPM", "revolutions/min")

print(temp.name, temp.unit)
print(rpm.name, rpm.unit)
````

**Expected Output:**
```
Temperature Celsius
RPM revolutions/min
```

### Example 3: Adding Methods (Behavior)

Methods are functions that belong to a class. They always take `self` as the first parameter, which gives them access to the object's attributes.

````python
class Sensor:
    """A sensor that collects readings."""

    def __init__(self, name, unit):
        self.name = name
        self.unit = unit
        self.readings = []   # starts empty

    def add_reading(self, value):
        """Add a new reading."""
        self.readings.append(value)

    def get_mean(self):
        """Return the average reading."""
        if not self.readings:
            return 0.0
        return sum(self.readings) / len(self.readings)

    def __str__(self):
        """Human-readable string representation."""
        count = len(self.readings)
        return "Sensor(" + self.name + ", " + str(count) + " readings)"


# Create and use
temp = Sensor("Temperature", "Celsius")
temp.add_reading(22.5)
temp.add_reading(23.1)
temp.add_reading(21.8)

print(temp)
print("Mean:", round(temp.get_mean(), 2), temp.unit)
print("All readings:", temp.readings)
````

**Expected Output:**
```
Sensor(Temperature, 3 readings)
Mean: 22.47 Celsius
All readings: [22.5, 23.1, 21.8]
```

### Example 4: Multiple Objects Have Separate State

This is one of the most important ideas in OOP: each object has its **own** copy of the attributes. Changing one object does NOT change another.

````python
temp = Sensor("Temperature", "C")
rpm = Sensor("RPM", "rpm")

temp.add_reading(22.5)
temp.add_reading(23.0)

rpm.add_reading(1500)
rpm.add_reading(1520)
rpm.add_reading(1510)

# Each has its OWN readings
print("temp readings:", temp.readings)
print("rpm  readings:", rpm.readings)
print()
print("temp mean:", round(temp.get_mean(), 1))
print("rpm  mean:", round(rpm.get_mean(), 1))
````

**Expected Output:**
```
temp readings: [22.5, 23.0]
rpm  readings: [1500, 1520, 1510]

temp mean: 22.8
rpm  mean: 1510.0
```

## Section 3: State vs Behavior

Every object has two aspects:

| Aspect | What it means | Example (Sensor) |
|--------|--------------|-------------------|
| **State** (attributes) | The data the object holds | `name`, `unit`, `readings` |
| **Behavior** (methods) | What the object can do | `add_reading()`, `get_mean()` |

**State** is stored in attributes (variables attached to `self`).
**Behavior** is defined by methods (functions inside the class).

Think of a robot:
- **State:** battery_level=87, position=(3,5), is_moving=True
- **Behavior:** move_forward(), turn_left(), pick_up()

The key insight: **methods can read and change the state**. The `add_reading` method changes the `readings` attribute. The `get_mean` method reads the `readings` attribute.

````python
# Demonstrating state change
s = Sensor("Vibration", "mm/s")
print("Before:", s.readings, "-- mean:", s.get_mean())

s.add_reading(0.5)
s.add_reading(1.2)
print("After: ", s.readings, "-- mean:", round(s.get_mean(), 2))

# The method changed the state!
print("Number of readings:", len(s.readings))
````

**Expected Output:**
```
Before: [] -- mean: 0.0
After:  [0.5, 1.2] -- mean: 0.85
Number of readings: 2
```

## Section 4: Procedural vs OOP -- Side by Side

---
### Procedural vs OOP: Managing Sensor Data

**Why is the OOP version better?**

1. **Encapsulation**: Data and functions that operate on it are bundled together. You cannot accidentally pass the wrong dictionary to the wrong function.
2. **Discoverability**: Type `s.` and your IDE shows you everything a Sensor can do.
3. **Safety**: In the procedural version, nothing stops you from writing `s["readnigs"]` (typo) -- you get a silent KeyError at runtime. With OOP, `s.readnigs` gives an immediate AttributeError.
4. **Reusability**: One class definition creates unlimited sensors, each with its own state.

**Procedural approach (what you did in CP1/CP2):**

````python
# PROCEDURAL: functions + dictionaries

def create_sensor(name, unit):
    return {"name": name, "unit": unit, "readings": []}

def add_reading(sensor, value):
    sensor["readings"].append(value)

def get_mean(sensor):
    r = sensor["readings"]
    return sum(r) / len(r) if r else 0

# Usage
s = create_sensor("Temp", "C")
add_reading(s, 22.5)
add_reading(s, 23.0)
print(get_mean(s))
````

**OOP approach (what we are learning now):**

````python
# OOP: class bundles data + functions

class Sensor:
    def __init__(self, name, unit):
        self.name = name
        self.unit = unit
        self.readings = []

    def add_reading(self, value):
        self.readings.append(value)

    def get_mean(self):
        return sum(self.readings) / len(self.readings) if self.readings else 0

# Usage
s = Sensor("Temp", "C")
s.add_reading(22.5)
s.add_reading(23.0)
print(s.get_mean())
````

## Section 5: Understanding `self`

`self` is the **most confusing part** of Python OOP for beginners. Let us demystify it.

When you write `temp.add_reading(22.5)`, Python translates it to `Sensor.add_reading(temp, 22.5)`. The object before the dot is automatically passed as the first argument -- that is `self`.

**Rule:** Every method in a class must have `self` as its first parameter. Inside the method, `self.something` accesses or creates an attribute on that specific object.

**Analogy:** `self` is like saying 'my' -- when a sensor says `self.name`, it means 'my name'. When you have two sensors, each one's `self` refers to itself.

````python
# Python translates method calls like this:
# temp.add_reading(22.5)  -->  Sensor.add_reading(temp, 22.5)

# We can prove it:
temp = Sensor("Temp", "C")

# Normal way
temp.add_reading(100)

# Explicit way (same thing!)
Sensor.add_reading(temp, 200)

print(temp.readings)  # both 100 and 200 are there
````

**Expected Output:**
```
[100, 200]
```

## Section 6: Common Mistakes

OOP introduces new error patterns. Let us see the most common ones and learn to fix them.

---
### Common Mistake: Forgetting `self` in method definition

The code below has a bug. Can you spot it before reading the fix?

````python
class Broken:
    def __init__(self, name):
        self.name = name

    def greet():  # BUG: missing self!
        print("Hello from " + self.name)

b = Broken("test")
try:
    b.greet()
except TypeError as e:
    print("ERROR:", e)
````

**What goes wrong:** Python passes the object as the first argument automatically. If you do not have `self` in the parameter list, Python tries to pass the object but there is no parameter to receive it. You get: `TypeError: greet() takes 0 positional arguments but 1 was given`.

**The fix:**

````python
class Fixed:
    def __init__(self, name):
        self.name = name

    def greet(self):  # FIXED: added self
        print("Hello from " + self.name)

f = Fixed("test")
f.greet()
````

---
### Common Mistake: Forgetting `self.` when accessing attributes

The code below has a bug. Can you spot it before reading the fix?

````python
class Broken2:
    def __init__(self, name):
        self.name = name
        self.count = 0

    def increment(self):
        count += 1  # BUG: this is a local variable, not self.count!

b = Broken2("test")
try:
    b.increment()
except UnboundLocalError as e:
    print("ERROR:", e)
````

**What goes wrong:** Without `self.`, Python thinks `count` is a local variable inside the method. It has never been assigned locally, so you get an `UnboundLocalError`. Always use `self.attribute` to access object state.

**The fix:**

````python
class Fixed2:
    def __init__(self, name):
        self.name = name
        self.count = 0

    def increment(self):
        self.count += 1  # FIXED: use self.count

f = Fixed2("test")
f.increment()
f.increment()
print("count:", f.count)
````

---
### Common Mistake: Creating attributes outside __init__

The code below has a bug. Can you spot it before reading the fix?

````python
class Risky:
    def __init__(self, name):
        self.name = name
        # Oops -- forgot to initialize self.data here

    def load(self):
        self.data = [1, 2, 3]  # created here instead

    def process(self):
        return sum(self.data)  # crashes if load() not called first!

r = Risky("test")
try:
    r.process()  # AttributeError!
except AttributeError as e:
    print("ERROR:", e)
````

**What goes wrong:** **Best practice:** Always initialize ALL attributes in `__init__`. Use `None`, `[]`, or a sensible default. This way, every object is in a valid state from the moment it is created.

**The fix:**

````python
class Safe:
    def __init__(self, name):
        self.name = name
        self.data = []  # FIXED: always initialize in __init__

    def load(self):
        self.data = [1, 2, 3]

    def process(self):
        if not self.data:
            print("No data loaded yet!")
            return 0
        return sum(self.data)

s = Safe("test")
print(s.process())  # safe: returns 0
s.load()
print(s.process())  # 6
````

---
### Debugging Tip: AttributeError: object has no attribute ...

This is the #1 OOP error. It means you tried to access an attribute that does not exist on the object. Common causes:

1. **Typo** in the attribute name (`self.naem` instead of `self.name`)
2. **Forgot to initialize** the attribute in `__init__`
3. **Forgot `self.`** when setting the attribute (wrote `name = x` instead of `self.name = x`)

**Fix:** Check `__init__` -- is the attribute there? Check spelling. Use `print(dir(obj))` to see all attributes an object has.

---
### Debugging Tip: TypeError: __init__() takes N positional arguments but M were given

This means you are passing the wrong number of arguments when creating an object.

```python
class Sensor:
    def __init__(self, name, unit):  # expects 2 args
        ...

s = Sensor("Temp")  # only 1 arg -> TypeError!
```

**Fix:** Count the parameters in `__init__` (excluding `self`). Pass exactly that many arguments.

### Design Decision: When to use a class vs a function

**Use a function** when:
- You have a single operation with no state to remember
- The operation is simple and self-contained
- Example: `def celsius_to_fahrenheit(temp): return temp * 9/5 + 32`

**Use a class** when:
- You have data AND operations that go together
- You need to remember state between calls
- You want to create multiple instances with different configurations
- Example: A `Sensor` that accumulates readings over time

**Rule of thumb:** If you find yourself passing the same dictionary to multiple functions, those functions and that dictionary probably want to be a class.

## Section 7: Putting It All Together -- Pipeline Preview

Let us build a tiny version of what your v3 pipeline will look like. This previews the next 13 weeks.

````python
class DataSource:
    """Loads raw data from a path."""

    def __init__(self, path):
        self.path = path
        self.raw_data = None

    def load(self):
        # Simulate loading
        self.raw_data = [
            {"id": 1, "value": 25.0, "status": "ok"},
            {"id": 2, "value": 30.0, "status": "ok"},
            {"id": 3, "value": -5.0, "status": "error"},
        ]
        print("Loaded " + str(len(self.raw_data)) + " rows from " + self.path)
        return self.raw_data


class Cleaner:
    """Cleans data by removing invalid rows."""

    def __init__(self, min_val=0, max_val=100):
        self.min_val = min_val
        self.max_val = max_val
        self.dropped = 0

    def clean(self, data):
        result = []
        self.dropped = 0
        for row in data:
            val = row.get("value")
            if isinstance(val, (int, float)) and self.min_val <= val <= self.max_val:
                result.append(row)
            else:
                self.dropped += 1
        print("Cleaned: kept " + str(len(result)) + ", dropped " + str(self.dropped))
        return result


class Analyzer:
    """Computes statistics on clean data."""

    def __init__(self, column="value"):
        self.column = column
        self.results = {}

    def analyze(self, data):
        values = [row[self.column] for row in data if self.column in row]
        if not values:
            self.results = {"count": 0}
            return self.results
        self.results = {
            "count": len(values),
            "mean": round(sum(values) / len(values), 2),
            "min": min(values),
            "max": max(values),
        }
        print("Analysis: " + str(self.results))
        return self.results


# Wire them together
source = DataSource("data/raw/sensors.csv")
cleaner = Cleaner(min_val=0, max_val=100)
analyzer = Analyzer("value")

raw = source.load()
clean = cleaner.clean(raw)
results = analyzer.analyze(clean)

print()
print("Each component has ONE job.")
print("Each component remembers its own state.")
print("Components are INDEPENDENT -- swap one without touching others.")
````

**Expected Output:**
```
Loaded 3 rows from data/raw/sensors.csv
Cleaned: kept 2, dropped 1
Analysis: {'count': 2, 'mean': 27.5, 'min': 25.0, 'max': 30.0}

Each component has ONE job.
Each component remembers its own state.
Components are INDEPENDENT -- swap one without touching others.
```
