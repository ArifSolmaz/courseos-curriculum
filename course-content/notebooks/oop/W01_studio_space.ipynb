### Step 1: Define Your Star Class

```python
class Star:
    """Represents a star in Lightcurve Explorer."""

    def __init__(self, star_id, name):
        self.star_id = star_id
        self.name = name
        self.data = []  # TODO: add appropriate attributes

    def add_data(self, value):
        """Add a data point."""
        self.data.append(value)

    def summary(self):
        """Return a summary of this star."""
        count = len(self.data)
        if count == 0:
            return "No data yet"
        mean = sum(self.data) / count
        return str(count) + " readings, mean=" + str(round(mean, 2))

    def __str__(self):
        return "Star(" + self.star_id + ")"


# Create instances
obj1 = Star("A001", "Primary")
obj2 = Star("B002", "Secondary")

obj1.add_data(42.0)
obj1.add_data(43.5)
obj2.add_data(100.0)

print(obj1, "->", obj1.summary())
print(obj2, "->", obj2.summary())
```
