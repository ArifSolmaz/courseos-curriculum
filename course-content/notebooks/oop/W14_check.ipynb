```python
import os, json

print("=== OOP Week 14 Universal Check ===")
print()
passed = 0
failed = 0

# Check 1: Core module exists
core_path = "src/project_template/core"
if os.path.exists(core_path):
    print("[PASS] " + core_path + "/ exists")
    passed += 1
else:
    print("[FAIL] " + core_path + "/ missing")
    failed += 1

# Check 2: Required exports
for f in ["data/cleaned/cleaned.csv", "reports/report.json"]:
    if os.path.exists(f) and os.path.getsize(f) > 0:
        print("[PASS] " + f)
        passed += 1
    else:
        print("[FAIL] " + f)
        failed += 1

# Check 3: Tests exist
if os.path.exists("tests"):
    test_files = [f for f in os.listdir("tests") if f.startswith("test_")]
    if test_files:
        print("[PASS] " + str(len(test_files)) + " test file(s) found")
        passed += 1
    else:
        print("[FAIL] No test files in /tests")
        failed += 1
else:
    print("[FAIL] /tests directory missing")
    failed += 1

print()
print(str(passed) + " passed, " + str(failed) + " failed")
```
