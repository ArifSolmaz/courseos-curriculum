## Section 1: What is an Architecture Freeze?

An **architecture freeze** means:

- The **public API** is locked -- no more changes to method signatures or return types
- The **component boundaries** are fixed -- no more moving classes between modules
- The **data contracts** are stable -- input/output schemas do not change

You CAN still:
- Fix bugs
- Improve performance
- Add new plugins (OCP!)
- Improve documentation

## Section 3: Interface Contracts

Each component has a **contract** -- a promise about what it accepts and what it returns.

````python
contracts = {
    "DataSource": {
        "method": "load()",
        "input": "path (set in __init__)",
        "output": "Dataset object",
        "errors": "SchemaError if columns missing",
    },
    "Cleaner": {
        "method": "clean(dataset)",
        "input": "Dataset object",
        "output": "NEW Dataset object (original unchanged)",
        "errors": "CleaningError if too much dropped",
    },
    "Analyzer": {
        "method": "analyze(dataset) or analyze(values)",
        "input": "Dataset or list of values",
        "output": "dict with analysis_summary key",
        "errors": "EmptyDataError if no data",
    },
    "Plotter": {
        "method": "plot(dataset, results)",
        "input": "Dataset + results dict",
        "output": "list of figure objects, saved to disk",
        "errors": "graceful degradation if matplotlib missing",
    },
    "Reporter": {
        "method": "export(dataset, results, figures)",
        "input": "Dataset + results + figure paths",
        "output": "dict of exported file paths",
        "errors": "IOError if disk full/permissions",
    },
}

print("=== Component Contracts ===")
for name, contract in contracts.items():
    print()
    print(name + ":")
    for key, val in contract.items():
        print("  " + key + ": " + val)
````

---
## Section 4: Automated Verification

````python
import os

def verify_architecture():
    """Automated architecture checks."""
    print("=== Automated Architecture Verification ===")
    passed = 0
    failed = 0

    # Check required output files
    files = [
        "data/cleaned/cleaned.csv",
        "reports/report.json",
    ]
    for f in files:
        exists = os.path.exists(f) and os.path.getsize(f) > 0
        status = "PASS" if exists else "FAIL"
        print("  [" + status + "] " + f)
        if exists:
            passed += 1
        else:
            failed += 1

    # Check core module
    core = "src/project_template/core/__init__.py"
    exists = os.path.exists(core)
    status = "PASS" if exists else "FAIL"
    print("  [" + status + "] " + core)
    if exists:
        passed += 1
    else:
        failed += 1

    # Check tests
    if os.path.exists("tests"):
        test_files = [f for f in os.listdir("tests") if f.startswith("test_")]
        has_tests = len(test_files) >= 3
        status = "PASS" if has_tests else "FAIL"
        print("  [" + status + "] " + str(len(test_files)) + " test files")
        if has_tests:
            passed += 1
        else:
            failed += 1
    else:
        print("  [FAIL] tests/ directory missing")
        failed += 1

    print()
    print(str(passed) + " passed, " + str(failed) + " failed")
    return failed == 0

verify_architecture()
````
