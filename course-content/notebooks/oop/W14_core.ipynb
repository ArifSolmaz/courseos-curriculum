## Section 2: Architecture Review

Let us review our architecture against the principles we learned.

````python
principles = [
    ("SRP", "Single Responsibility",
     "Each component (DataSource, Cleaner, Analyzer, Plotter, Reporter) has ONE job"),
    ("OCP", "Open/Closed",
     "New analyzers added via AnalyzerFactory without changing existing code"),
    ("Composition", "Has-A relationships",
     "Pipeline HAS components; components do not inherit from Pipeline"),
    ("Strategy", "Interchangeable algorithms",
     "Cleaning and analysis strategies are swappable via config"),
    ("Factory", "Object creation from config",
     "AnalyzerFactory creates objects from string names"),
    ("Exceptions", "Domain-specific errors",
     "PipelineError hierarchy with context-rich messages"),
]

print("=== Architecture Principles Review ===")
for abbr, name, description in principles:
    print()
    print(abbr + " -- " + name)
    print("  " + description)
````

## Section 6: Architecture Comparison: v2 vs v3

````python
comparison = {
    "Code organization": {
        "v2": "Functions in one file",
        "v3": "Classes in separate modules",
    },
    "Adding features": {
        "v2": "Modify existing functions",
        "v3": "Add new classes (OCP)",
    },
    "Error handling": {
        "v2": "Generic ValueError/print",
        "v3": "Domain exceptions with context",
    },
    "Testing": {
        "v2": "Manual print checking",
        "v3": "pytest with fixtures and parametrize",
    },
    "Configuration": {
        "v2": "Hardcoded values",
        "v3": "Config-driven, factory pattern",
    },
    "Reusability": {
        "v2": "Copy-paste between projects",
        "v3": "Import and configure",
    },
}

print("=== v2 vs v3 Comparison ===")
for aspect, versions in comparison.items():
    print()
    print(aspect + ":")
    print("  v2: " + versions["v2"])
    print("  v3: " + versions["v3"])
````

**Expected Output:**
```

Code organization:
  v2: Functions in one file
  v3: Classes in separate modules
```
